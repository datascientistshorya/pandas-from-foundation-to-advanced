# Missing-value detection and assessment

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "age": [25, np.nan, 35, 42, np.nan, 29, 38, 45],
    "income": [30000, 50000, np.nan, 70000, 45000, np.nan, 60000, 80000],
    "city": ["Delhi", "Mumbai", None, "Pune", "Delhi", None, "Mumbai", "Pune"],
    "purchased": [1, 0, 1, None, 0, 1, None, 1]
})
df
```

#### Q1. Detect missing values
Use isna() to identify missing values in every column. Display the resulting Boolean DataFrame.

```python
df.isna().sum()
```

```python
df.isna()
```

#### Q2. Count missing values
Calculate the number of missing values in each column, ordered from most missing to least missing.

```python
df.isna().sum().sort_values(ascending=False)
```

#### Q3. Calculate missing-value percentages
Calculate the percentage of missing values in each column. Round the results to two decimal places.

```python
(df.isna().mean()*100).round(2)
```

#### Q4. Identify incomplete rows
Display all rows containing at least one missing value using isna() and any().

```python
df[df.isna().any(axis=1)]
```

#### Q5. Identify complete rows
Return only rows with no missing values in any column. Explain how this differs from removing rows that have missing values in just one selected column.

```python
df[~df.isna().any(axis=1)]
```

#### Q6. Drop rows with missing values
Use dropna() to remove rows containing any missing value. Store the result in a new DataFrame and report the number of remaining rows.

```python
df_clean=df.dropna()
df_clean
```

```python
df_clean.shape[0]
```

#### Q7. Drop columns with missing values
Use dropna(axis=1) to retain only columns with no missing values. Explain why this can be risky on real-world datasets.

```python
df_clean2 =df.dropna(axis=1)
df_clean2
```

#### Q8. Drop based on selected columns
Remove rows where age or income is missing, but allow missing values in city and purchased. Use subset.

```python
df_clean3=df.dropna(axis=0,subset=['age','income'])
df_clean3
```

#### Q9. Assess missingness by column
Create a summary DataFrame with three columns: missing_count, missing_percentage, and data_type. Each row should correspond to a column in the original dataset.

```python
summary=pd.DataFrame(
    {'missing_count':df.isna().sum(),
    'missing_percentage':df.isna().mean()*100,
    'data_type':df.dtypes}
)
summary
```

#### Q10. Mini-project — Missing-data audit
Create a reusable function missing_data_report(data) that:
1. Reports each column's missing count and percentage.
2. Includes each column's data type.
3. Sorts columns by missing percentage descending.
4. Returns the summary as a DataFrame without modifying the original dataset.
5. Test it on the dataset above.
6. Explain why a data analyst should audit missingness before choosing an imputation strategy.

```python

def missing_data_report(df):
    data=pd.DataFrame(
        {'missing_count':df.isna().sum(),
        'missing%':(df.isna().mean())*100,})
    data=data.sort_values(by='missing%',ascending =False)


    return data
data_summary=missing_data_report(df)
data_summary
```
