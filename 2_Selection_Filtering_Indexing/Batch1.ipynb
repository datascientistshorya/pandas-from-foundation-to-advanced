#### Q8. Use isin() and between()

1. Select customers from Delhi or Mumbai using isin().
2. Select customers aged between 25 and 35 using between().
3. Select customers from Delhi or Pune who spend more than 1,500.
4. Select customers whose IDs are in [102, 104, 106].

```python
df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106],
    "city": ["Delhi", "Mumbai", "Pune", "Delhi", "Jaipur", "Mumbai"],
    "age": [22, 35, 28, 41, 30, 25],
    "spend": [1200, 2500, 1800, 4000, 2200, 1500]
})
df
```

```python
df[df['city'].isin(['Delhi','Mumbai'])]
```

```python
df[df['age'].between(25,35)]
```

```python
#Select customers from Delhi or Pune who spend more than 1,500.
df[(df['city'].isin(['Delhi','Pune'])) & df['spend']>1500]
```

```python
#Select customers whose IDs are in [102, 104, 106].
df[(df['customer_id'].isin([102, 104, 106]))]
```

#### Q10. Mini-project — Customer Data Selection
Create this DataFrame:
df = pd.DataFrame({    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108],    "age": [22, 35, 28, 41, 30, 25, 45, 33],    "monthly_spend": [1200, 2500, 1800, 4000, 2200, 1500, 5000, 3100],    "visits": [4, 8, 5, 9, 6, 3, 12, 7],    "city": ["Delhi", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune", "Delhi", "Jaipur"],    "churn": [0, 1, 0, 1, 1, 0, 0, 1]})



Complete the following:
1. Display only customer_id, monthly_spend, and churn.
2. Select customers spending more than 2,000.
3. Select customers aged at least 30 with 7 or more visits.
4. Select customers from Delhi or Mumbai.
5. Select customers who churned and spent at least 2,000.
6. Select customers who did not churn and visited fewer than 6 times.
7. Select rows at positions 1, 3, and 5 using iloc.
8. Display age and churn for the first four rows using iloc.
9. Select customers whose age is between 25 and 40, inclusive.
10. Write a short explanation of how these selection and filtering operations help with ML data analysis.

```python
df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "age": [22, 35, 28, 41, 30, 25, 45, 33],
    "monthly_spend": [1200, 2500, 1800, 4000, 2200, 1500, 5000, 3100],
    "visits": [4, 8, 5, 9, 6, 3, 12, 7],
    "city": ["Delhi", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune", "Delhi", "Jaipur"],
    "churn": [0, 1, 0, 1, 1, 0, 0, 1]
})
df
```

```python
#Display only customer_id, monthly_spend, and churn.
df[['customer_id','monthly_spend','churn']]
```

```python
#Select customers spending more than 2,000.
df[df['monthly_spend']>2000]
```

```python
#Select customers aged at least 30 with 7 or more visits.
df[(df['age']>=30) & (df['visits']>=7)]
```

```python
#Select customers from Delhi or Mumbai.
df[df['city'].isin(['Delhi','Mumbai'])]
```

```python
#Select customers who churned and spent at least 2,000.
df[(df['churn']==1) & df['monthly_spend']>=2000]
```

```python
#Select customers who did not churn and visited fewer than 6 times.
df[(df['churn']==0) & (df['visits']<=6)]
```

```python
#Select rows at positions 1, 3, and 5 using iloc.
df.iloc[[1,3,5]]
```

```python
#Display age and churn for the first four rows using iloc.
df.iloc[0:4,[1,5]]
```

```python
#Select customers whose age is between 25 and 40, inclusive.
df[df['age'].between(25,40)]
```
