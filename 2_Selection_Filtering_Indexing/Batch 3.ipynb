# Batch 3- Index Alignment, Sorting and Ranking

```python
import pandas as pd

df = pd.DataFrame({
    "employee": ["Asha", "Rahul", "Neha", "Vikram", "Priya", "Aman"],
    "department": ["IT", "Sales", "IT", "HR", "Sales", "HR"],
    "salary": [60000, 45000, 75000, 50000, 45000, 65000],
    "experience": [3, 2, 5, 4, 2, 6],
    "performance_score": [85, 78, 92, 80, 88, 92]
})
df
```

#### Q1. Sort by one column
Sort employees by salary in ascending order.

```python
df.sort_values(by='salary')
```

#### Q2. Sort by multiple columns
Sort by department alphabetically, then by salary in descending order within each department.

```python
df.sort_values(by=['department','salary'],ascending=[False,True])
```

#### Q3. Reset the index
Sort the DataFrame by salary in descending order and reset the index, dropping the old index.

```python
df_sorted=df.sort_values(by='salary',ascending=False)
df_sorted
```

```python
df_sorted.reset_index(drop=True)
```

#### Q6. Rank salaries
Create a new column named salary_rank that ranks salaries from highest to lowest, with the highest salary receiving rank 1.

```python
df['rank_salary']=df['salary'].rank(ascending=False).astype(int)
df
```

#### Q7. Handle tied ranks
Rank performance_score from highest to lowest using the dense method. Explain how dense ranking differs from the default ranking method when scores are tied.

```python
df['performance_rank']=df['performance_score'].rank(ascending=False,method='min').astype(int)
df
```

#### Q8. Find the top performers

Calculate s1 + s2. Explain why Pandas aligns values by index labels rather than simply adding values at matching positions, and why some results are missing.

#### Q10. Mini-project — Employee ranking system
Using the original DataFrame:
1. Sort employees by performance score descending and salary descending.
2. Assign a dense performance rank.
3. Filter employees with performance scores of at least 85.
4. Return employee, department, salary, performance_score, and performance_rank.
5. Reset the index.
6. Explain how ranking and sorting can help an analyst identify high-performing employees without losing track of tied scores.

```python
df = pd.DataFrame({
    "employee": ["Asha", "Rahul", "Neha", "Vikram", "Priya", "Aman"],
    "department": ["IT", "Sales", "IT", "HR", "Sales", "HR"],
    "salary": [60000, 45000, 75000, 50000, 45000, 65000],
    "experience": [3, 2, 5, 4, 2, 6],
    "performance_score": [85, 78, 92, 80, 88, 92]
})
df
```

```python
#Sort employees by performance score descending and salary descending.
df=df.sort_values(by=['performance_score','salary'],ascending=[False,True])
df
```

```python
#Assign a dense performance rank.
df['performance_rank']=df['performance_score'].rank(method='min',ascending=False).astype(int)
df
```

```python
#Filter employees with performance scores of at least 85.
df=df[df['performance_score']>=85]
df
```

```python
df=df.reset_index(drop=True)
df
```
