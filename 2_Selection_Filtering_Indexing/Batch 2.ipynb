# Batch 2 - Advanced Filtering, Boolean Masks and Conditional Selection

```python
import pandas as pd

df = pd.DataFrame({
    "customer": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "age": [22, 35, 28, 42, 31, 25, 38, 45],
    "purchase_amount": [500, 1500, 800, 2000, 1200, 700, 1800, 2500],
    "city": [
        "Delhi", "Mumbai", "Delhi", "Pune",
        "Mumbai", "Pune", "Delhi", "Mumbai"
    ],
    "membership": [
        "Basic", "Premium", "Basic", "Premium",
        "Premium", "Basic", "Premium", "Basic"
    ]
})
df
```

#### Q1. Multiple conditions with AND
Select customers whose age is greater than 30 and whose purchase_amount is greater than 1000.
Return only customer, age, and purchase_amount.

```python
df[(df['age']>30) & (df['purchase_amount']>1000)][['customer','age','purchase_amount']]
```

#### Q2. Multiple conditions with OR
Select customers who live in Delhi or have a purchase amount greater than 2000.
Return all columns.

```python
df[(df['city']=='Delhi') | (df['purchase_amount']>2000)]
```

#### Q3. Combining AND and OR
Select customers who meet either condition:
- They live in Mumbai and have Premium membership.
- Their age is greater than 40.
Return customer, age, city, and membership.

```python
df[((df['city']=='Mumbai') & (df['membership']=='Premium')
|
(df['age']>40))][['customer','age','city','membership']]
```

#### Q4. Filtering with isin()
Select customers who live in either Delhi or Pune.
Use isin() rather than multiple OR conditions.

```python
df[(df['city'].isin(['Delhi','Pune']))]
```

#### Q5. Excluding values
Select all customers who are not Basic members.
Solve this using the ~ operator and isin()

```python
df[~(df['membership']).isin(['Basic'])]
```

#### Q6. Range-based filtering
Select customers whose age is between 25 and 40, inclusive.
Solve this using Boolean conditions, then solve it again using between().

```python
df[(df['age'].between(25,40))]
```

```python
df[(df['age']>=25) & (df['age']<=40)]
```

#### Q7. Conditional selection with query()
Use df.query() to select customers whose age is greater than 30 and whose purchase amount is at least 1500.
Return customer, age, and purchase_amount.

```python
df.query('age>30 and purchase_amount>=1500')[['customer','age','purchase_amount']]
```

#### Q8. Negating a Boolean condition
Select customers who either:
- Do not live in Mumbai, or
- Have a purchase amount less than 1000.
Use Boolean masks. Be careful with operator precedence.

```python
df[~((df['city']=='Mumbai') | (df['purchase_amount']<1000))]
```

#### Q9. Missing values and Boolean masks
Create this DataFrame:
data = pd.DataFrame({    "customer": ["A", "B", "C", "D", "E"],    "age": [25, None, 35, 42, None],    "income": [30000, 50000, None, 70000, 45000]})



Select rows where age is not missing and income is greater than 40000. Explain why notna() is useful.

```python
data = pd.DataFrame({
    "customer": ["A", "B", "C", "D", "E"],
    "age": [25, None, 35, 42, None],
    "income": [30000, 50000, None, 70000, 45000]
})
data
```

```python
data[(data['age'].notna()) & (data['income']>40000)]
```
