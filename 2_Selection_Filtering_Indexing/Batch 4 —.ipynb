#### Batch 4 — Advanced Selection and Indexing Mini-Projects

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "age": [22, 35, 28, 42, 31, 25, 38, 45],
    "city": [
        "Delhi", "Mumbai", "Delhi", "Pune",
        "Mumbai", "Pune", "Delhi", "Mumbai"
    ],
    "orders": [2, 8, 4, 10, 6, 3, 9, 12],
    "revenue": [500, 2500, 1200, 4000, 1800, 700, 3200, 5000],
    "satisfaction": [3.5, 4.8, 4.0, 4.9, 4.2, 3.2, 4.7, 4.9]
})
df
```

#### Q1. Conditional column selection
Select customers whose revenue exceeds 2,000 and whose satisfaction score is at least 4.5. Return only customer_id, revenue, and satisfaction.

```python
df[(df['revenue']>2000) & (df['satisfaction']>=4.5)][['customer_id','revenue','satisfaction']]
```

#### Q2. Advanced Boolean filtering
Select customers who live in Delhi or Mumbai, have at least 5 orders, and are younger than 40. Use properly parenthesized Boolean masks.

```python
df[(df['city'].isin(['Delhi','Mumbai'])) & (df['orders']>=5) & (df['age']<40)]
```

#### Q3. loc with conditional selection
Use .loc to select rows where revenue is at least 1,800 and update their satisfaction score by adding 0.1. Cap the updated score at 5.0.

```python
mask=df['revenue']>=1800
df.loc[mask,'satisfaction']=(df.loc[mask,'satisfaction']+0.1).clip(upper=5.0)
df.loc[mask]
```

#### Q4. iloc with row and column positions
Retrieve rows at positions 1, 3, and 5, and columns at positions 0, 1, and 4. Explain the difference between label-based and position-based selection.

```python
df.iloc[[1,3,5],[0,1,4]]
```

#### Q6. Multi-condition ranking
Rank customers by revenue in descending order, using the min method for ties. Explain how tied ranks are assigned.

```python
df['revenue_rank']=df['revenue'].rank(method='min',ascending=False).astype(int)
df
```

#### Q7. Missing-value filtering
Create a copy of the DataFrame and set the revenue for customer 103 and satisfaction for customer 106 to NaN. Select rows where both values are present using notna().

```python
df_copy=df.copy()
df_copy.loc[df_copy['customer_id'] == 103, 'revenue'] = np.nan
df_copy.loc[df_copy['customer_id'] == 106, 'satisfaction'] = np.nan
df_copy
```

```python
df_copy[(df_copy['revenue'].notna()) & (df_copy['satisfaction'].notna())]
```

#### Q8. Top-N selection
Use nlargest() to retrieve the four customers with the highest revenue. Then sort those four customers by age in ascending order.

```python
high_rev=df.nlargest(4,'revenue').sort_values(by='age')
high_rev
```

#### Q9. Safe conditional updates
Create a copy of the DataFrame. Add a high_value column that is True when revenue is at least 2,500 and False otherwise. Then increase orders by 1 only for high-value customers, using .loc.

```python
df_high=df.copy()
df_high['high_value']= df_high['revenue']>=2500
df_high
```

```python
mask=df_high['high_value']==True
df_high.loc[mask,'orders']=df_high.loc[mask,'orders']+1
df_high
```

#### Q10. Mini-project — Customer targeting
Build a customer-targeting table that:
1. Selects customers aged 25–40, inclusive.
2. Requires revenue of at least 1,200 and satisfaction of at least 4.0.
3. Returns customer_id, age, city, revenue, and satisfaction.
4. Sorts by revenue descending.
5. Adds a revenue rank using dense ranking.
6. Resets the index.
7. Explains how this table could help identify customers for a marketing campaign and why selection criteria should be chosen carefully.

```python
df
```

```python
#Selects customers aged 25–40, inclusive.
final=df[df['age'].between(25,40)]
final
```

```python
#Requires revenue of at least 1,200 and satisfaction of at least 4.0.
final=final.query("revenue>=1200 and satisfaction>=4.0")
final
```

```python
#Returns customer_id, age, city, revenue, and satisfaction.
final=final.drop(columns=['orders','revenue_rank'])
final
```

```python
#Sorts by revenue descending.
final=final.sort_values(by='revenue', ascending = False)
final
```

```python
#Adds a revenue rank using dense ranking.
final['revenue_rank']=final['revenue'].rank(method='min',ascending = False).astype(int)
final
```

```python
#Resets the index.
final=final.reset_index(drop=True)
final
```
