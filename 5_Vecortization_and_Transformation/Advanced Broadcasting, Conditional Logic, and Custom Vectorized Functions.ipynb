# Advanced Broadcasting, Conditional Logic, and Custom Vectorized Functions

```python

import pandas as pd
import numpy as np

data = {
    'customer_id': range(101, 116),
    'name': [
        'Aarav', 'Priya', 'Rahul', 'Sneha', 'Vikram',
        'Ananya', 'Rohan', 'Neha', 'Karan', 'Isha',
        'Arjun', 'Meera', 'Dev', 'Kavya', 'Aditi'
    ],
    'city': [
        'Delhi', 'Mumbai', 'Bangalore', 'Delhi', 'Mumbai',
        'Bangalore', 'Pune', 'Delhi', 'Mumbai', 'Pune',
        'Bangalore', 'Delhi', 'Pune', 'Mumbai', 'Bangalore'
    ],
    'age': [
        24, 31, 28, 42, 35,
        26, 39, 22, 45, 33,
        29, 37, 41, 27, 30
    ],
    'income': [
        25000, 45000, 70000, 18000, 85000,
        55000, 32000, 0, 95000, 42000,
        -5000, 60000, 28000, 75000, np.nan
    ],
    'spending': [
        8000, 18000, 30000, 5000, 45000,
        22000, 12000, 7000, 110000, 25000,
        15000, 35000, -2000, 27000, 16000
    ],
    'savings': [
        5000, 15000, 30000, 8000, 40000,
        22000, 12000, 0, 50000, 18000,
        -1000, 26000, 9000, 35000, np.nan
    ],
    'monthly_expenses': [
        12000, 22000, 35000, 10000, 40000,
        28000, 18000, 6000, 45000, 25000,
        15000, 30000, 14000, 32000, 20000
    ]
}

df = pd.DataFrame(data)

df
```

#### Q1. City-based bonus using vectorization
Create a city_bonus column using these rules:
- Delhi: 5,000
- Mumbai: 7,000
- Bangalore: 6,000
- All other cities: 2,000

```python
conditions=[
    df['city']=='Delhi',
    df['city']=='Mumbai',
    df['city']=='Bangalore'
    
]
bonus=[5000,7000,6000]
df['city_bonus']=np.select(conditions,bonus,default=2000)
df
```

#### Q2. Scale income using broadcasting
Create adjusted_income by increasing each customer's income by 10%.

Then create income_after_expenses by subtracting a fixed expense of 5,000 from each adjusted income.

```python
df['adjusted_income']=df['income']*1.1
df['income_after_expenses']=df['adjusted_income']-5000
df
```

#### Q3. Create spending tiers
Create a spending_tier column using these rules:
- Spending below 10,000: Low
- Spending from 10,000 to below 25,000: Medium
- Spending of 25,000 or above: High

```python
conditions=[
    df['spending']<10000,
    df['spending'].between(10000,25000),
    df['spending']>25000
]
labels=['Low','Medium','High']
df['spending_tier']=np.select(conditions,labels)
df
```

#### Q4. Conditional tax calculation
Create a tax column using these rules:
- Income below 30,000: 0% tax
- Income from 30,000 to below 70,000: 10% tax
- Income of 70,000 or above: 20% tax
The tax is calculated on the customer's full income, not just the amount above each threshold.

```python
conditions=[
    df['income']<30000,
    df['income'].between(30000,70000),
    df['income']>70000
    
]
tax_applied=[0,10,20]
df['tax']=np.select(conditions,tax_applied)

df['tax_on_income']=(df['income']*df['tax'])/100
df
```

#### Q5. Calculate savings contributions
Create monthly_contribution equal to 15% of savings for customers whose savings are at least 20,000. For everyone else, assign 5% of savings.

```python
df['monthly_contribution']=(df['savings']*0.15).where(df['savings']>=20000,df['savings']*0.05)
df
```

#### Q6. City-average comparison
Calculate the average income for each city and create a Boolean column, above_city_average, indicating whether each customer's income is greater than their city's average income.

```python
df['above_city_average']=df['income']>df.groupby('city')['income'].transform('mean')
df
```

#### Q7. Clip extreme spending values
Create spending_clipped by limiting spending to a minimum of 0 and a maximum of 100,000.

```python
df['spending_clipped']=df['spending'].clip(lower=0,upper=100000)
df
```

#### Q8. Safe vectorized division
Create spending_ratio by dividing spending by income.

If income is zero or negative, the result must be NaN. Avoid division-by-zero problems and unnecessary row-wise operations.

```python
df['spending_ratio']=df['spending']/df['income'].where(df['income']>0,np.nan)
df
```

#### Q9. Validate a transformation pipeline
Create valid_income by replacing zero and negative income values with NaN.

Then create income_category:
- Below 25,000: Low
- From 25,000 to below 60,000: Medium
- 60,000 or above: High

```python
df['valid_income'] = df['income'].where(df['income'] > 0, np.nan)
conditions=[
    df['valid_income']<25000,
    df['valid_income'].between(25000,60000),
    df['valid_income']>60000
]
labels=['Low','Medium','High']
df['income_category']=np.select(conditions,labels,default='n/a')
df

```
