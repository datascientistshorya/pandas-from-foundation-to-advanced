#### Q1. Revenue performance by city
Create a DataFrame named city_performance grouped by city, containing:
- order_count: number of orders
- valid_sales_count: non-missing sales count
- total_sales: sum of valid sales
- average_sales: mean sales
- median_sales: median sales
- total_quantity: total quantity sold
- average_discount: average discount percentage
Sort the result by total_sales in descending order.

```python
city_performance=df.groupby('city').agg(
    order_count=('order_id','count'),
    valid_sales=('sales','count'),
    total_sales=('sales','sum'),
    avg_sales=('sales','mean'),
    meadian_sales=('sales','median'),
    total_quantity=('quantity','sum'),
    avg_discount=('discount_pct','mean')
)
city_performance
```

Sort by total sales from highest to lowest.

#### Q3. Order-status analysis
Create status_report, grouped by order_status, with:
- Order count
- Total sales
- Average sales
- Total quantity
- Unique customers
Calculate the percentage of orders in each status:
\[
\text{Order Share (\%)} =
\frac{\text{Orders in Status}}{\text{All Orders}}\times100
\]

Store this as order_share_pct.

```python
status_report=df.groupby('order_status').agg(
    order_count=('order_id','count'),
    total_sales=('sales','sum'),
    avg_sales=('sales','mean'),
    total_quantity=('quantity','sum'),
    unique_customers=('customer_id','unique')
    
)
status_report['order_share_pct'] = (
    status_report['order_count'] / df['order_id'].count()
) * 100
status_report
```

#### Q4. Channel-level business KPIs
Create channel_report, grouped by channel, with:
- Order count
- Total sales
- Average sales
- Median sales
- Unique customers
- Average discount
Then calculate each channel's contribution to total recorded sales as sales_contribution_pct.

```python
channel_report=df.groupby('channel').agg(
    order_count=('order_id','count'),
    total_sales=('sales','sum'),
    avg_sales=('sales','mean'),
    median_sales=('sales','median'),
    unique_customers=('customer_id','unique'),
    avg_discount=('discount_pct','mean')
)
channel_report
```

```python
channel_report['sales_contribution_pct']=(channel_report['total_sales']/df['sales'].sum())*100
channel_report
```

#### Q5. City and category performance
Group by both city and category.

Create city_category_report containing:
- Order count
- Total sales
- Average sales
- Total quantity
- Average discount
- Unique customers
Sort by city and then total sales in descending order within each city.

Identify the highest-sales category for every city.

```python
city_category_report=df.groupby(['city','category']).agg(
    order_count=('order_id','count'),
    total_sales=('sales','sum'),
    avg_sales=('sales','mean'),
    median_sales=('sales','median'),
    unique_customers=('customer_id','unique'),
    avg_discount=('discount_pct','mean')
)
city_category_report = city_category_report.sort_values(
    by=['city', 'total_sales'],
    ascending=[True, False]
)
city_category_report
```
