#### Q1. Transform: City-average sales
Create a new column city_avg_sales containing the average sales for the corresponding city on every row.

Requirements:
- Use groupby() with transform().
- Preserve the original row count and index.
- Missing sales should be handled using pandas' normal mean behavior.

```python
df['city_avg_sales']=df.groupby('city')['sales'].transform('mean')
df
```

#### Q2. Transform: Deviation from city average
Create sales_vs_city_avg by subtracting the city-average sales from each order's sales.

Formula:

sales_vs_city_avg = sales - city_avg_sales

Requirements:
- Use vectorized operations.
- Preserve missing sales as missing.
- Interpret positive and negative values.

```python
df['sales_vs_city_avg']=df['sales']-df['city_avg_sales']
df
```

#### Q3. Multiple transformations by category
Create these columns:
- category_avg_sales: category-average sales.
- category_total_quantity: total quantity within the category.
- category_order_count: number of rows within the category.
Use transform() so each result aligns with the original DataFrame.

```python
df['category_avg_sales']=df.groupby('category')['sales'].transform('mean')
df['category_total_quantity']=df.groupby('category')['quantity'].transform('sum')
df['category_order_count']=df.groupby('category')['order_id'].transform('count')
```

#### Q4. Filter high-performing cities
Return only cities whose total valid sales exceed 100,000.

Requirements:
- Use groupby() and filter().
- Keep the original rows belonging to qualifying cities.
- Store the filtered DataFrame in high_sales_cities.

```python
dff = df.dropna(subset=['sales']).copy()
dff
```

```python
high_sales_cities = dff.groupby('city').filter(
    lambda group: group['sales'].sum() > 100000
)
high_sales_cities
```

#### Q5. Filter categories by order volume
Keep only categories with at least eight orders.

Requirements:
- Use groupby() and filter().
- Count rows, not non-missing sales.
- Store the result in high_volume_categories.

```python
high_volume_categories = df.groupby('category').filter(
    lambda group: len(group) >= 8
)
high_volume_categories
```

#### Q6. Group-wise percentage contribution
Create category_sales_share showing each order's sales as a percentage of the total valid sales for its category.

Formula:

sales / category_total_sales * 100

Requirements:
- Use transform() for the category total.
- Handle missing sales.
- Avoid division by zero.

```python
category_total_sales = df.groupby('category')['sales'].transform('sum')

df['category_sales_share'] = (
    df['sales'] / category_total_sales.where(category_total_sales != 0, np.nan)
) * 100
df
```

#### Q7. Rank orders within each city
Create sales_rank_in_city ranking sales from highest to lowest within each city.

Requirements:
- Use groupby() and rank().
- Highest sales should receive rank 1.
- Use method="dense" for ties.
- Missing sales should retain missing ranks.

```python
df['sales_rank_in_city']=df.groupby('city')['sales'].rank(method='dense',ascending=False)
df
```

#### Q8. Calculate within-city cumulative sales
Create cumulative_sales_by_city using cumulative sales within each city.

Requirements:
- Sort orders by order_id before calculating the cumulative sum.
- Use groupby() and cumsum().
- Preserve the original DataFrame unchanged.
- Explain how missing sales affect the cumulative result.

```python
dff = df.sort_values('order_id').copy()
dff['cumulative_sales_by_city']=df.groupby('city')['sales'].cumsum()
dff
```

#### Q9. Compare each order with its category median
Create above_category_median as a Boolean column indicating whether each order's sales exceed its category's median sales.

Requirements:
- Calculate the category median using transform("median").
- Return missing results for rows with missing sales.
- Do not use row-wise apply().

```python
df['category_median']=df.groupby('category')['sales'].transform('median')
df['above_category_median']=df['sales']>df['category_median']
df
```
