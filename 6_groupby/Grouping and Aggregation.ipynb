# Grouping and Aggregation

```python
import pandas as pd
import numpy as np

data = {
    "order_id": range(1001, 1021),
    "customer_id": [
        201, 202, 203, 204, 205,
        206, 207, 208, 209, 210,
        201, 202, 203, 204, 205,
        206, 207, 208, 209, 210
    ],
    "city": [
        "Delhi", "Mumbai", "Delhi", "Pune", "Mumbai",
        "Delhi", "Bangalore", "Pune", "Mumbai", "Bangalore",
        "Delhi", "Mumbai", "Delhi", "Pune", "Mumbai",
        "Delhi", "Bangalore", "Pune", "Mumbai", "Bangalore"
    ],
    "category": [
        "Electronics", "Clothing", "Electronics", "Furniture",
        "Clothing", "Electronics", "Clothing", "Furniture",
        "Electronics", "Clothing", "Furniture", "Clothing",
        "Electronics", "Electronics", "Furniture",
        "Clothing", "Electronics", "Furniture", "Clothing",
        "Electronics"
    ],
    "sales": [
        15000, 4500, 22000, 12000, 6500,
        18000, 5500, 14000, 25000, 7000,
        11000, 8000, 19000, 16000, 9000,
        6000, 21000, 13000, np.nan, 17000
    ],
    "quantity": [
        2, 3, 4, 1, 5,
        2, 4, 2, 3, 5,
        1, 4, 2, 3, 2,
        6, 3, 2, 4, 1
    ],
    "discount_pct": [
        5, 10, 0, 15, 10,
        5, 0, 20, 5, 10,
        10, 5, 0, 10, 15,
        0, 5, 10, 5, 0
    ],
    "order_status": [
        "Completed", "Completed", "Cancelled", "Completed",
        "Completed", "Completed", "Cancelled", "Completed",
        "Completed", "Completed", "Completed", "Cancelled",
        "Completed", "Completed", "Completed", "Cancelled",
        "Completed", "Completed", "Completed", "Cancelled"
    ]
}

df = pd.DataFrame(data)

df
```

#### Q1. Total Sales by City
Calculate the total sales for each city.

Requirements:
1. Group the DataFrame by city.
2. Calculate the sum of sales.
3. Store the result in city_sales.
4. Sort the results in descending order of total sales.
Concepts: groupby(), sum(), sort_values().

```python
city_sales=(df.groupby('city')['sales'].sum().sort_values(ascending=False))
city_sales
```

#### Q2. Average Sales by Product Category
Calculate the average sales for each product category.

Requirements:
1. Group by category.
2. Calculate the mean of sales.
3. Store the result in category_avg_sales.
4. Round the results to two decimal places.

```python
category_avg_sales=round((df.groupby('category')['sales'].mean()),2)
category_avg_sales
```

#### Q3. Number of Orders by City
Count the number of orders in each city.

Requirements:
1. Group by city.
2. Count the rows in each group using size().
3. Store the result in orders_by_city.
4. Sort from highest to lowest order count.

```python
orders_by_city=df.groupby('city')['city'].size().sort_values(ascending=False)
orders_by_city
```

#### Q4. Non-Missing Sales Count by City
Calculate the number of non-missing sales values in each city.

Requirements:
1. Group by city.
2. Count non-null values in sales.
3. Store the result in valid_sales_count.

```python
valid_sales_count=df.groupby('city')['sales'].count()
valid_sales_count
```

#### Q5. Total Quantity Sold by Category
Calculate the total quantity sold for each product category.

Requirements:
1. Group by category.
2. Sum the quantity column.
3. Store the result in category_quantity.
4. Identify the category with the highest total quantity.

```python
category_quantity=df.groupby('category')['quantity'].sum()
category_quantity
```

```python
category_quantity.idxmax()
```

#### Q6. Minimum and Maximum Sales by City
Calculate the minimum and maximum sales values for each city.

Requirements:
1. Group by city.
2. Calculate the minimum sales.
3. Calculate the maximum sales.
4. Store both results in a DataFrame called city_sales_range.

```python
city_max=df.groupby('city')['sales'].max()
city_min=df.groupby('city')['sales'].min()
city_sales_range=pd.DataFrame([city_max,city_min])
city_sales_range.index = ['city_max', 'city_min']
city_sales_range
```

#### Q7. Median Sales by Product Category
Calculate the median sales for each category.

Requirements:
1. Group by category.
2. Calculate the median of sales.
3. Store the result in category_median_sales.
4. Explain briefly why the median can differ from the mean.

#### Q8. Unique Customers by City
Calculate the number of distinct customers represented in each city.

Requirements:
1. Group by city.
2. Count unique customer_id values.
3. Store the result in unique_customers_by_city

```python
unique_customers_by_city=df.groupby('city')['customer_id'].unique()
unique_customers_by_city
```

#### Q9. Multiple Sales Aggregations
Calculate multiple sales metrics for each product category.

Requirements:
1. Group by category.
2. Calculate the count of non-missing sales values.
3. Calculate total sales.
4. Calculate average sales.
5. Calculate median sales.
6. Store the result in category_sales_summary.
7. Use agg() to calculate the metrics together.

```python
category_sales_summary=df.groupby('category')['sales'].agg(['mean','median','sum'])
category_sales_summary
```

#### Q10. Business Summary by Order Status
Create a business summary grouped by order_status.

Requirements:
1. Group by order_status.
2. Calculate the number of orders.
3. Calculate total sales.
4. Calculate average sales.
5. Calculate total quantity.
6. Store the result in status_summary.
7. Compare the completed and cancelled orders based on the resulting metrics.

```python
status_summary = df.groupby('order_status').agg(
    total_sales=('sales', 'sum'),
    average_sales=('sales', 'mean'),
    order_count=('sales', 'count'),
    total_quantity=('quantity', 'sum')
)

status_summary
```
