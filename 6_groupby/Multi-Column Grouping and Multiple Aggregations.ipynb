# Multi-Column Grouping and Multiple Aggregations

```python
import pandas as pd
import numpy as np

data = {
    "order_id": range(2001, 2026),
    "customer_id": [
        301, 302, 303, 304, 305,
        306, 307, 308, 309, 310,
        301, 302, 303, 304, 305,
        306, 307, 308, 309, 310,
        311, 312, 313, 314, 315
    ],
    "city": [
        "Delhi", "Delhi", "Mumbai", "Mumbai", "Pune",
        "Pune", "Bangalore", "Bangalore", "Delhi", "Mumbai",
        "Delhi", "Delhi", "Mumbai", "Mumbai", "Pune",
        "Pune", "Bangalore", "Bangalore", "Delhi", "Mumbai",
        "Pune", "Delhi", "Bangalore", "Mumbai", "Pune"
    ],
    "category": [
        "Electronics", "Clothing", "Electronics", "Furniture",
        "Clothing", "Electronics", "Clothing", "Furniture",
        "Furniture", "Clothing", "Electronics", "Furniture",
        "Clothing", "Electronics", "Furniture", "Clothing",
        "Electronics", "Clothing", "Electronics", "Furniture",
        "Electronics", "Clothing", "Furniture", "Clothing",
        "Electronics"
    ],
    "sales": [
        25000, 5000, 18000, 14000, 6500,
        22000, 7500, 16000, 12000, 9000,
        30000, 11000, 8000, 24000, 15000,
        5500, 28000, 6000, np.nan, 13000,
        19000, 4500, 21000, 10000, 17000
    ],
    "quantity": [
        3, 2, 2, 1, 4,
        3, 5, 2, 1, 3,
        4, 2, 1, 3, 2,
        2, 4, 2, 3, 2,
        2, 1, 3, 2, 4
    ],
    "discount_pct": [
        5, 10, 0, 15, 10,
        5, 0, 20, 10, 5,
        0, 15, 10, 5, 20,
        0, 5, 10, 5, 15,
        10, 0, 5, 10, 0
    ],
    "order_status": [
        "Completed", "Completed", "Completed", "Cancelled",
        "Completed", "Completed", "Completed", "Cancelled",
        "Completed", "Completed", "Completed", "Cancelled",
        "Completed", "Completed", "Completed", "Cancelled",
        "Completed", "Completed", "Completed", "Completed",
        "Cancelled", "Completed", "Completed", "Cancelled",
        "Completed"
    ]
}

df = pd.DataFrame(data)

df
```

#### Q1. Total sales by city and category
Calculate total sales for every combination of city and category.

Requirements:
1. Group by both columns.
2. Sum sales.
3. Store the result in city_category_sales.
4. Sort by total sales in descending order.

```python
city_category_sales=df.groupby(['city','category'])['sales'].sum().sort_values(ascending=False)
city_category_sales
```

#### Q2. Average sales by city and category
Calculate average sales for each city-category combination.

Requirements:
1. Group by city and category.
2. Calculate mean sales.
3. Store the result in city_category_avg.
4. Round to two decimal places.

```python
city_category_avg=round(df.groupby(['city','category'])['sales'].mean(),2)
city_category_avg
```

#### Q3. Order count by city and status
Calculate the number of orders for each combination of city and order status.

Requirements:
1. Group by city and order_status.
2. Count rows using size().
3. Store the result in city_status_count.
4. Make sure orders with missing sales are still counted.

```python
city_status_count=df.groupby(['city','order_status']).size()
city_status_count
```

#### Q4. Multiple aggregations by city and category
For every city-category combination, calculate:
- Total sales
- Average sales
- Median sales
- Total quantity
- Number of non-missing sales values
Store the result in city_category_summary.

```python
city_category_summary=(df.groupby(['city','category'])).agg(
    total_sales=('sales','sum'),
    avg_sales=('sales','mean'),
    median_sales=('sales','median'),
    total_quantity=('quantity','sum'),
    count_sales=('sales','count'))

city_category_summary
```

#### Q5. Find the best-performing category in each city
Identify the category with the highest total sales in each city.

Requirements:
1. Calculate total sales by city and category.
2. Find the category with the maximum sales within each city.
3. Store the final result in top_category_by_city.
4. Return one top category per city when there are no ties; document how you handle ties.

```python
category_by_city=df.groupby(['city','category'])['sales'].sum()
category_by_city
```

```python
top_category_by_city = (
    category_by_city
    .loc[category_by_city.groupby(level='city').idxmax()]
)

top_category_by_city
```

#### Q6. Unique customers by city and category
Count distinct customers for every city-category combination.

Requirements:

#### Q8. Average discount by city and category
Calculate the average discount_pct for each city-category combination.

#### Q9. Completed-order sales by city and category
Analyze completed orders only.

#### Q10. Mini-project — Multi-dimensional business report
Build a report grouped by city, category, and order_status.

Include:
1. order_count: number of orders.
2. valid_sales_count: number of non-missing sales values.
3. total_sales: sum of sales.
4. average_sales: mean sales.
5. median_sales: median sales.
6. total_quantity: sum of quantity.
7. average_discount: average discount percentage.
8. unique_customers: distinct customers.
Store the result in business_report.

Then:
- Sort the report by total_sales in descending order.
- Inspect the first 10 rows.
- Check for missing values.
- Explain two business insights supported by the results.

```python
business_report=df.groupby(['city','category','order_status']).agg(
    order_count=('order_id','count'),
    valid_sales_count=('sales','count'),
    total_sales=('sales','sum'),
    avg_sales=('sales','mean'),
    median_sales=('sales','median'),
    total_quantity=('quantity','sum'),
    avg_discount=('discount_pct','mean'),
    unique_customers=('customer_id','unique')
)
business_report
```

```python
business_report=business_report.sort_values(by='total_sales',ascending=False)
business_report
```
