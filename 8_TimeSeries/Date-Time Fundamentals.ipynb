# Date-Time Fundamentals

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'order_id': range(1001, 1013),
    'customer_id': [101, 102, 103, 101, 104, 105,
                    102, 106, 103, 104, 107, 101],
    'order_date': [
        '2026-01-05 09:30:00',
        '2026-01-12 14:15:00',
        '2026-02-03 11:00:00',
        '2026-02-18 16:45:00',
        '2026-03-07 10:20:00',
        '2026-03-21 18:30:00',
        '2026-04-02 08:45:00',
        '2026-04-15 12:10:00',
        '2026-05-09 15:00:00',
        '2026-05-22 19:20:00',
        '2026-06-11 13:35:00',
        '2026-06-25 17:50:00'
    ],
    'delivery_date': [
        '2026-01-08', '2026-01-17', '2026-02-08',
        '2026-02-23', '2026-03-10', '2026-03-29',
        '2026-04-05', '2026-04-21', '2026-05-15',
        '2026-05-27', '2026-06-16', '2026-06-30'
    ],
    'sales': [2500, 1800, 3200, 4200, 1500, 3800,
              2100, 4500, 2700, 3300, 1900, 5000],
    'status': [
        'Completed', 'Cancelled', 'Completed', 'Completed',
        'Completed', 'Cancelled', 'Completed', 'Completed',
        'Cancelled', 'Completed', 'Completed', 'Completed'
    ]
})

# Convert date columns to datetime
df['order_date'] = pd.to_datetime(df['order_date'])
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
df
```

```python
df.dtypes
```

#### Q2. Extract the year

Create a year column from order_date.

```python
df['year']=df['order_date'].dt.year
df
```

#### Q3. Extract month information

Create two columns:
- order_month: numeric month.
- month_name: full month name.

```python
df['order_month']=df['order_date'].dt.month
df['order_month_name']=df['order_date'].dt.month_name()
df
```

#### Q4. Extract day components

Create day, day_name, and quarter columns from order_date.

```python
df['day']=df['order_date'].dt.day
df['day_name']=df['order_date'].dt.day_name()
df['quarter']=df['order_date'].dt.quarter
df
```

#### Q5. Filter a date range

Select all orders placed between March 1 and April 30, 2026, inclusive.

```python
sel_orders=df[df['order_date'].between('2026-03-01','2026-04-30')]
sel_orders
```

#### Q6. Filter by month

Select all orders placed in May 2026 using a datetime-based condition.

```python
sel_order=df[df['order_month']==5]
sel_order
```

#### Q7. Calculate delivery duration

Create a delivery_days column representing the number of days between order_date and delivery_date. Use the date portion of order_date, not its time of day.

```python
df['delivery_days']=(df['delivery_date']-df['order_date']).dt.days
df
```

#### Q8. Find delayed deliveries

Identify orders whose delivery took more than five calendar days. Display their order IDs and delivery durations.

```python
sel_orders=df[df['delivery_days']>5]
sel_orders
```

#### Q9. Find the weekend orders

Filter orders placed on Saturday or Sunday using the date-time accessors.

```python
sel_orders=df[df['order_date'].dt.dayofweek.isin([5,6])]
sel_orders
```

#### Q10. Business analysis by month

Create a monthly report containing:
- order_count
- total_sales
- average_sales
- unique_customers
- cancelled_orders
Group by month in chronological order.

```python
report=df.groupby('order_month').agg(
    order_count=('order_id','count'),
    total_sales=('sales','sum'),
    avg_sales=('sales','mean'),
    unique_customers=('customer_id','nunique'),
    cancelled_orders=('status',lambda x: (x == 'Cancelled').sum())
)
report
```
