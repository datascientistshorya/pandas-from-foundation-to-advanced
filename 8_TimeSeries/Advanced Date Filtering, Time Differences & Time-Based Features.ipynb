# Advanced Date Filtering, Time Differences & Time-Based Features

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'order_id': range(2001, 2016),
    'customer_id': [
        101, 102, 101, 103, 104, 102, 105, 101,
        103, 106, 104, 105, 102, 107, 101
    ],
    'order_date': [
        '2026-01-05', '2026-01-19', '2026-02-05',
        '2026-02-27', '2026-03-02', '2026-03-15',
        '2026-03-31', '2026-04-05', '2026-04-20',
        '2026-05-01', '2026-05-17', '2026-06-01',
        '2026-06-15', '2026-06-29', '2026-06-30'
    ],
    'sales': [
        1200, 2500, 1800, 3200, 900, 4100, 1500,
        2700, 3500, 2200, 4800, 1300, 2900, 1700, 5200
    ],
    'status': [
        'Completed', 'Completed', 'Cancelled', 'Completed',
        'Completed', 'Completed', 'Cancelled', 'Completed',
        'Completed', 'Completed', 'Cancelled', 'Completed',
        'Completed', 'Completed', 'Completed'
    ]
})

df['order_date'] = pd.to_datetime(df['order_date'])
df
```

#### Q1. Month start and month end

Create month_start and month_end columns representing the beginning and end of each order's calendar month.

```python
df['month_start'] = df['order_date'].dt.to_period('M').dt.start_time
df['month_end'] = df['order_date'].dt.to_period('M').dt.end_time

df
```

#### Q2. Quarter identification

Create a quarter_label column such as 2026-Q1. Use the order date to derive both the year and quarter

```python
df['quarter_label'] = (
    df['order_date'].dt.year.astype(str)
    + '-Q'
    + df['order_date'].dt.quarter.astype(str)
)

df
```

#### Q3. Previous and next dates

Create columns for the previous calendar day and the next calendar day relative to each order date.

```python
df['previous_day'] = df['order_date'] - pd.Timedelta(days=1)
df['next_day'] = df['order_date'] + pd.Timedelta(days=1)

df
```

#### Q4. Date offsets

Calculate the date 7 days after each order and the date 30 days before each order using Pandas date offsets or timedeltas.

```python
df['date_offset']=df['order_date']+pd.Timedelta(days=7)
df
```

#### Q5. Business-day calculation

Calculate the number of business days between each order date and seven calendar days after it. Exclude Saturdays and Sundays using NumPy's busday_count.

```python
df['business_days'] = np.busday_count(
    df['order_date'].values.astype('datetime64[D]'),
    df['date_offset'].values.astype('datetime64[D]')
)

df
```

#### Q6. Month-to-date filtering

For a fixed reference date of June 20, 2026, filter orders from June 1 through June 20, inclusive.

```python
sel_orders=df[df['order_date'].between('2026-06-01','2026-06-20')]
sel_orders
```

#### Q7. Trailing 30-day window

For the same reference date, select orders in the trailing 30-calendar-day window, including June 20 but excluding dates before May 22. Explain your boundary conditions.

```python
reference_date = pd.Timestamp('2026-06-20')
start_date = reference_date - pd.Timedelta(days=29)

sel_orders = df[
    df['order_date'].between(start_date, reference_date, inclusive='both')
]

sel_orders
```

#### Q8. Customer purchase recency

For each customer, calculate the number of days since their most recent order as of June 30, 2026

```python
reference_date = pd.Timestamp('2026-06-30')

customer_report = (
    df.groupby('customer_id', as_index=False)
      .agg(last_order_date=('order_date', 'max'))
)

customer_report['days_since_last_order'] = (
    reference_date - customer_report['last_order_date']
).dt.days

customer_report
```

#### Q9. First and latest purchase

Create a customer-level DataFrame containing each customer's first purchase date, latest purchase date, and total number of orders.

```python
report=df.groupby('customer_id').agg(
    first_purchase=('order_date','min'),
    latest_purchase=('order_date','max'),
    NO_of_orders=('order_id','count')
)
report
```

#### Q10. Customer activity segmentation

Using June 30, 2026, as the reference date, categorize each customer based on their latest purchase:
- Active: 0–30 days ago
- At Risk: 31–60 days ago
- Inactive: more than 60 days ago
Build a customer-level report with customer_id, last_order_date, days_since_last_order, and activity_status.

```python

reference_date = pd.Timestamp('2026-06-30')

customer_report = (
    df.groupby('customer_id', as_index=False)
      .agg(last_order_date=('order_date', 'max'))
)

customer_report['days_since_last_order'] = (
    reference_date - customer_report['last_order_date']
).dt.days


df = df.merge(
    customer_report[['customer_id', 'days_since_last_order']],
    on='customer_id',
    how='left'
)
```

```python
conditions=[
    df['days_since_last_order'].between(0,30),
    df['days_since_last_order'].between(31,60),
    df['days_since_last_order']>60
    
]
labels=['Active','At Risk','Inactive']
df['activity_status']=np.select(conditions,labels)
df
```
