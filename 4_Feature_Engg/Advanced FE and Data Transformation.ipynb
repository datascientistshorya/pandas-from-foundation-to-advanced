# Advanced Feature Engineering and Data Transformation

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'customer_id': range(301, 313),
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai', 'Pune',
             'Delhi', 'Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Pune'],
    'category': ['Electronics', 'Clothing', 'Electronics', 'Food',
                 'Clothing', 'Food', 'Food', 'Electronics',
                 'Clothing', 'Electronics', 'Food', 'Clothing'],
    'income': [30000, 45000, 55000, 22000, 38000, 28000,
               60000, 52000, 41000, 35000, 48000, 25000],
    'spending': [12000, 18000, 25000, 9000, 15000, 11000,
                 30000, 22000, 17000, 14000, 21000, 10000],
    'orders': [2, 4, 6, 1, 3, 2, 7, 5, 4, 3, 6, 1],
    'tenure_months': [6, 12, 24, 3, 9, 6, 30, 18, 12, 9, 24, 3]
})
df
```

#### Q1. Customer spending rank

Create a spending_rank column that ranks customers by spending from highest to lowest. The highest spender should receive rank 1.

```python
df['spending_rank']=df['spending'].rank(ascending=False)
df
```

#### Q2. City-level spending rank

Rank customers by spending within each city. Create a city_spending_rank column using groupby() and rank().

```python
df['city_spending_rank']=df.groupby('city')['spending'].rank(ascending=False)
df
```

#### Q3. Category-level average

Create a category_avg_spending column containing the average spending for each customer's product category. Every customer should receive the average for their category.

```python
df['category_avg_spending']=df.groupby('category')['spending'].transform('mean')
df
```

#### Q4. Deviation from city average

Create a city_spending_deviation column by subtracting each customer's city-average spending from their own spending.

```python
df['city_spending_deviation'] = (
    df['spending']
    - df.groupby('city')['spending'].transform('mean')
)
df
```

#### Q5. Group-based normalization

Create a city_income_minmax column that normalizes income separately within each city to a range of 0–1. Handle groups where all income values are identical.

```python
city_min=df.groupby('city')['income'].transform('min')
city_max=df.groupby('city')['income'].transform('max')
city_range=city_max-city_min
df['city_minmax']=(df['income']-city_min)/city_range.where(city_range!=0).fillna(0.0)
df
```

#### Q6. Percentile-based segmentation

Use pd.qcut() to divide customers into four spending groups: Q1, Q2, Q3, and Q4. Handle potential duplicate bin edges appropriately.

```python
df['spending_quartile'] = pd.qcut(
    df['spending'].rank(method='first'),
    q=4,
    labels=['Q1', 'Q2', 'Q3', 'Q4']
)
df
```

#### Q7. Cumulative spending

Sort customers by spending from highest to lowest and create a cumulative_spending column showing the running total of spending. Preserve the original customer identifiers.

```python
df_sorted = df.sort_values(
    by='spending',
    ascending=False
).copy()

df_sorted['cumulative_spending'] = (
    df_sorted['spending'].cumsum()
)
df_sorted
```

#### Q8. City-level spending share

Create a city_spending_share column showing each customer's spending as a percentage of their city's total spending.

```python
city_total = df.groupby('city')['spending'].transform('sum')

df['city_spending_share'] = (
    df['spending'] / city_total
) * 100
df
```

#### Q9. Relative customer value

Create a relative_customer_value column by dividing each customer's spending by the average spending in their city. Explain how to interpret values above and below 1.

```python
city_avg = df.groupby('city')['spending'].transform('mean')
df['relative_customer_value']=df['spending']/city_avg
df
```

Greater than 1: Customer spends more than the city average.



Equal to 1: Customer spends exactly the city average.



Less than 1: Customer spends less than the city average.
