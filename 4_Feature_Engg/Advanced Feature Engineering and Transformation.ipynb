# Advanced Feature Engineering and Transformation

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'customer_id': range(201, 211),
    'age': [19, 25, 34, 42, 51, 63, 29, 38, 47, 55],
    'income': [18000, 25000, 32000, 45000, 55000,
               80000, 0, 38000, 62000, 48000],
    'spending': [5000, 12000, 18000, 22000, 30000,
                 45000, 0, 15000, 35000, 27000],
    'orders': [1, 2, 4, 3, 5, 8, 0, 2, 7, 4],
    'tenure_months': [2, 6, 12, 18, 24, 48, 0, 9, 36, 30],
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai',
             'Delhi', 'Pune', 'Mumbai', 'Delhi', 'Pune']
})
df
```

#### Q1. Log transformation

Create a log_income column using np.log1p() to reduce the skewness of income. Explain why log1p() is safer than np.log() when income contains zero.

```python
df['log_income']=np.log1p(df['income'])
df
```

#### Q2. Log transformation of spending

Create a log_spending column using np.log1p(). Verify that all transformed values are finite.

```python
df['log_spending']=np.log1p(df['spending'])
df
```

#### Q3. Age binning

Use pd.cut() to divide customers into these age groups: 18–25, 26–35, 36–45, 46–55, and 56+. Store the groups in a new column called age_band.

```python
bins = [17, 25, 35, 45, 55, float('inf')]

labels = ['18–25', '26–35', '36–45', '46–55', '56+']

df['age_band'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels
)
df
```

#### Q4. Income segmentation

Create an income_segment column with the following categories:
- Below 25,000: Low
- 25,000–49,999: Medium
- 50,000–74,999: High
- 75,000 and above: Very High

```python
conditions=[
    df['income']<25000,
    df['income'].between(25000,49999),
    df['income'].between(50000,74999),
    df['income']>=75000
  
]
labels=['Low','Medium','High','Very High']
df['income_segment']=np.select(conditions,labels,default='n/a')
df
```

#### Q5. Spending-to-income ratio

Create spending_ratio, expressed as a percentage of income. Handle zero income safely and explain why replacing the undefined ratio with zero could be misleading.

```python
df['spending_ratio']=np.where(
    df['income']>0,
    (df['spending']/df['income'])*100,np.nan)
df
```

#### Q6. Customer eligibility

Create an eligible_for_premium Boolean column. A customer qualifies if their income is at least 50,000, their tenure is at least 12 months, and they have placed at least three orders.

```python
df['eligible_for_premium']=np.where(
    (df['income']>50000) & (df['tenure_months']>=12),
    'Yes','No')
df
```

#### Q7. Customer activity classification

Create activity_status using these rules:
- Inactive: zero orders
- New: one or two orders
- Active: three to five orders
- Highly Active: more than five orders

```python
conditions=[
    df['orders']==0,
    df['orders'].isin([1,2]),
    df['orders'].between(3,5),
    df['orders']>5
   ]
labels=['Inactive','New','Active','Highly Active']

df['activity_status']=np.select(conditions,labels,default='N/A')
df
```

#### Q8. Spending per order

Create spending_per_order by dividing spending by orders. Safely handle customers with zero orders and explain the resulting missing values.

```python
df['spending_per_order']=np.where(
    df['orders']>0,
    round(df['spending']/df['orders'],2),np.nan)
df
```

#### Q9. Combined feature engineering

Create three new features:
- income_per_tenure_month: income divided by tenure months, safely handling zero tenure.
- high_value_customer: True if income is at least 50,000 and spending is at least 30,000.
- customer_lifetime_activity: orders multiplied by tenure months.

```python
df['income_per_tenure_month']=np.where(
    df['income']>0,
    round(df['income']/df['tenure_months'],2),np.nan
)
df
```

high_value_customer: True if income is at least 50,000 and spending is at least 30,000.

```python
df['high_value_customer']=(df['income']>50000) & (df['spending']>=30000)
df
```

customer_lifetime_activity: orders multiplied by tenure months

```python
df['customer_lifetime_activity']=df['orders']*df['tenure_months']
df
```
