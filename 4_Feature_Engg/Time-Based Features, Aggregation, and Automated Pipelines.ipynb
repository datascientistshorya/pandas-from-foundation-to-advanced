# Time-Based Features, Aggregation, and Automated Pipelines

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'customer_id': range(401, 413),
    'city': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Pune',
             'Delhi', 'Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Pune'],
    'signup_date': pd.to_datetime([
        '2025-01-10', '2025-02-15', '2025-03-20',
        '2025-04-05', '2025-05-12', '2025-06-18',
        '2025-07-22', '2025-08-09', '2025-09-14',
        '2025-10-01', '2025-11-11', '2025-12-20'
    ]),
    'last_order_date': pd.to_datetime([
        '2026-01-10', '2026-02-20', '2026-03-15',
        '2026-01-25', '2026-04-12', '2026-05-18',
        '2026-02-22', '2026-06-09', '2026-03-14',
        '2026-07-01', '2026-05-11', '2026-08-20'
    ]),
    'income': [30000, 45000, 22000, 55000, 38000, 28000,
               60000, 52000, 41000, 35000, 48000, 25000],
    'spending': [12000, 18000, 9000, 25000, 15000, 11000,
                 30000, 22000, 17000, 14000, 21000, 10000],
    'orders': [2, 4, 1, 6, 3, 2, 7, 5, 4, 3, 6, 1]
})
df
```

#### Q1. Extract date features

Create signup_year, signup_month, signup_day, and signup_weekday from signup_date using Pandas datetime accessors.

```python
df['signup_year']=df['signup_date'].dt.year
df['signup_month']=df['signup_date'].dt.month
df['signup_day']=df['signup_date'].dt.day
df['signup_weekday']=df['signup_date'].dt.weekday
df
```

#### Q2. Calculate customer tenure

Create tenure_days as the number of days between last_order_date and signup_date. Identify any negative tenure values.

```python
df['tenure_days']=(df['last_order_date']-df['signup_date']).dt.days
df
```

#### Q3. Calculate recency

Using 2026-10-19 as the reference date, create days_since_last_order. Explain what a higher value indicates.

```python
df['days_since_last_order']=(pd.Timestamp('2026-10-19')-df['last_order_date']).dt.days
df
```

#### Q4. Create customer activity categories

Create recency_status using these rules:
- 0–90 days: Recent
- 91–180 days: Warming
- More than 180 days: Dormant

```python
conditions=[
    df['days_since_last_order'].between(0,90),
    df['days_since_last_order'].between(91,180),
    df['days_since_last_order']>180
    
]
labels=['Recent','Warming','Dormant']
df['recency_status']=np.select(conditions,labels,default='N/A')
df
```

#### Q5. Create spending efficiency features

Create spending_per_order by dividing spending by orders. Handle zero orders safely.

```python
df['spending_per_order']=df['spending']/df['orders'].where(df['orders']!=0,np.nan)
df
```

#### Q6. Calculate city-level metrics

Create city_avg_income and city_avg_spending using groupby() and transform('mean'). Each customer should receive the average for their city.

```python
df['city_avg_income']=df.groupby('city')['income'].transform('mean')
df
```

#### Q7. Create deviation features

Create income_vs_city_avg by subtracting city-average income from each customer's income. Create spending_vs_city_avg similarly.

```python
df['income_vs_city_avg ']=(df.groupby('city')['income'].transform('mean')-df['income'])
df
```

#### Q8. Create a composite customer score

Calculate customer_score using these normalized features:
- Income: 40% weight
- Spending: 35% weight
- Orders: 25% weight

```python
# Normalize each feature to a 0–1 range
features = ['income', 'spending', 'orders']

for col in features:
    min_val = df[col].min()
    max_val = df[col].max()

    if max_val != min_val:
        df[f'{col}_normalized'] = (
            (df[col] - min_val) / (max_val - min_val)
        )
    else:
        df[f'{col}_normalized'] = 0.0

# Calculate the composite customer score
df['customer_score'] = (
    df['income_normalized'] * 0.40 +
    df['spending_normalized'] * 0.35 +
    df['orders_normalized'] * 0.25
)

df
```

#### Q9. Create a rule-based customer segment

Create customer_segment using the following rules:
- High Value: income at least 50,000 and orders at least 5
- Regular: income at least 30,000 and orders at least 2
- Low Activity: all remaining customers
Ensure the conditions are evaluated in the correct order.

```python
conditions=[
    ((df['income']>=50000) & (df['orders']>=5)),
    ((df['income']>=30000) & (df['orders']>=2))
]
labels=['High Value','Regular']

df['customer_segment']=np.select(conditions,labels,default='Low Activity')
df
```
