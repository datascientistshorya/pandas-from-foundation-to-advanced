#### Q6. Renaming Columns
Consider:
df = pd.DataFrame({    "Customer Age": [25, 30, 35],    "Monthly Income": [30000, 45000, 60000],    "Churn Status": [0, 1, 0]})



Rename the columns to:
- customer_age
- monthly_income
- churn
Write code to perform the renaming and display the updated column names.
Explain why consistent column naming is useful in an ML project.

```python
df = pd.DataFrame({
    "Customer Age": [25, 30, 35],
    "Monthly Income": [30000, 45000, 60000],
    "Churn Status": [0, 1, 0]
})

df=df.rename(columns={'Customer Age': 'customer_age',
                        'Monthly Income':'monthly_income',
                     'Churn Status':'churn'})
df.columns
```

#### Q8. Missing Values and Inspection
Consider:
df = pd.DataFrame({    "age": [22, None, 28, 35],    "salary": [25000, 30000, None, 45000],    "department": ["Sales", "IT", "IT", None]})



Write code to:
a. Count missing values in each column.
b. Count the total missing values in the entire DataFrame.
c. Calculate the percentage of missing values in each column.
d. Explain why you should inspect missing values before deciding how to handle them.
Do not impute or remove any values.

```python
df = pd.DataFrame({
    "age": [22, None, 28, 35],
    "salary": [25000, 30000, None, 45000],
    "department": ["Sales", "IT", "IT", None]
})

df.isnull().sum()
```

```python
df_total_null=df.isnull().sum().sum()
df_total_null
```

```python
round((df.isnull().sum()/df.isnull().sum().sum())*100,2)
```
