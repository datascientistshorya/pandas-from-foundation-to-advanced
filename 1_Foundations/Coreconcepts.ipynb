#### Q3. Creating and Removing Columns
Given:
df = pd.DataFrame({    "quantity": [2, 3, 5],    "unit_price": [100, 200, 150],    "product": ["A", "B", "C"]})



Write code to:
a. Create a new column named total_price by multiplying quantity by unit_price.
b. Create a column named high_value that is True when total_price is greater than 500 and False otherwise.
c. Remove the product column from the DataFrame.
d. Display the resulting DataFrame.

```python
df = pd.DataFrame({
    "quantity": [2, 3, 5],
    "unit_price": [100, 200, 150],
    "product": ["A", "B", "C"]
})
df['total_price']=df['quantity']*df['unit_price']
df['high_value']=df['total_price']>500
df=df.drop(columns='product')
df
```

#### Q6. Resetting and Understanding Indexes
Given:
df = pd.DataFrame({    "name": ["Aman", "Priya", "Rahul"],    "score": [80, 90, 70]}, index=[10, 20, 30])



Answer the following:
a. What is the index of this DataFrame?
b. Write code to reset the index while retaining the original index as a column.
c. Write code to reset the index and discard the old index.
d. Explain why resetting an index might be useful after combining or transforming datasets.

```python
df = pd.DataFrame({
    "name": ["Aman", "Priya", "Rahul"],
    "score": [80, 90, 70]
}, index=[10, 20, 30])
#10, 20 ,30 are the indexes
df
```

```python
df=df.reset_index()
df
```

```python
df=df.drop(columns='index')
df
```

#### Resetting an index is useful after combining or transforming datasets because the original index may become duplicated, non-sequential, or no longer meaningful.

#### Q10. Mini-Project — Initial Customer Data Analysis
Create this DataFrame:
df = pd.DataFrame({    "customer_id": [101, 102, 103, 104, 105, 106],    "age": [22, 35, 29, 41, 35, 22],    "monthly_spend": [1200, 2500, 1800, 3200, 2500, 1200],    "visits": [4, 8, 5, 9, 8, 4],    "churn": [0, 1, 0, 1, 1, 0]})



Write a complete solution that:
1. Displays the DataFrame's shape and data types.
2. Reports the number of unique values in each column.
3. Displays the frequency of each churn class.
4. Calculates mean, median and standard deviation for monthly_spend.
5. Creates a spend_per_visit column by dividing monthly_spend by visits.
6. Sorts customers by monthly_spend in descending order.
7. Checks for completely duplicated rows.
8. Groups the data by churn and calculates average monthly_spend and visits.
9. Prints a short interpretation of one finding from the results.

```python
df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106],
    "age": [22, 35, 29, 41, 35, 22],
    "monthly_spend": [1200, 2500, 1800, 3200, 2500, 1200],
    "visits": [4, 8, 5, 9, 8, 4],
    "churn": [0, 1, 0, 1, 1, 0]
})

print('Shape: ',df.shape, '\nDatatypes:\n',df.dtypes)
```

```python
df.nunique()
```

```python
df['churn'].value_counts(normalize=True)*100
```

```python
#Calculates mean, median and standard deviation for monthly_spend
df['monthly_spend'].agg(['mean','median','std'])
```

```python
#Creates a spend_per_visit column by dividing monthly_spend by visits.

df['spend_per_visit']=round(df['monthly_spend']/df['visits'],2)
df
```

```python
#Sorts customers by monthly_spend in descending order.
df_sorted=df.sort_values(by='monthly_spend',ascending=False)
df_sorted
```

```python
#Checks for completely duplicated rows.
df[df.duplicated()]
```

```python
df.duplicated().sum()
```

```python
#Groups the data by churn and calculates average monthly_spend and visits.

round(df.groupby('churn')[['monthly_spend','visits']].mean(),2)
```
