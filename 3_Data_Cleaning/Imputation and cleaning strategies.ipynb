# Imputation and cleaning strategies

#### Q1. Fill missing values with a constant
Fill missing values in the city column with "Unknown". Store the result in a new DataFrame without modifying df.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "age": [25, np.nan, 35, 42, np.nan, 29, 38, 45],
    "income": [30000, 50000, np.nan, 70000, 45000, np.nan, 60000, 80000],
    "city": ["Delhi", "Mumbai", None, "Pune", "Delhi", None, "Mumbai", "Pune"],
    "purchased": [1, 0, 1, np.nan, 0, 1, np.nan, 1]
})
df
```

```python
df['city']=df['city'].fillna('unknown')
df
```

#### Q2. Mean imputation
Fill missing values in age with the mean age. Calculate the mean first and then use fillna().
Explain why mean imputation can be sensitive to outliers.

```python
df['age']=df['age'].fillna(round(df['age'].mean(),2))
df
```

#### Q3. Median imputation
Fill missing values in income with the median income. Explain why median imputation may be more appropriate than mean imputation for skewed income data.

```python
df['income']=df['income'].fillna(df['income'].median())
df
```

#### Q4. Mode imputation
Fill missing values in city using the most frequent city in the dataset. Use mode() and fillna().

```python
df['city']=df['city'].replace('unknown',df['city'].mode()[0])
df
```

#### Q10. Mini-project — Reusable imputation function
Write a function called impute_customer_data(data) that:
1. Creates a copy of the input DataFrame.
2. Fills missing numeric values using the median of each numeric column.
3. Fills missing categorical values using the mode of each categorical column.
4. Handles a categorical column containing only missing values without crashing.
5. Returns the cleaned DataFrame.
6. Tests the function on df and verifies that the original dataset remains unchanged.

```python
def impute_customer_data(data):
    df=data.copy()
    df['age']=df['age'].fillna(df['age'].median())
    df['income']=df['income'].fillna(df['income'].median())
    df['city']=df['city'].fillna(df['city'].mode()[0])

    return df

cleaned=impute_customer_data(df)
cleaned
    
```
