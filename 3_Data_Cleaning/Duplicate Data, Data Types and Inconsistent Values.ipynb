# Duplicate Data, Data Types and Inconsistent Values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "customer_id": [101, 102, 103, 102, 104, 105, 106, 106],
    "name": ["Asha", "rahul", "NEHA", "rahul",
             "Priya ", "Aman", "Riya", "Riya"],
    "age": ["25", " thirty", "35", " thirty",
            "29", "42", "28", "28"],
    "city": ["Delhi", "mumbai", "DELHI", "mumbai",
             " Pune ", "Delhi", None, None],
    "income": [30000, 50000, 45000, 50000,
               -1000, 70000, np.nan, np.nan]
})
df
```

#### Q1. Detect duplicate rows
Use duplicated() to identify complete duplicate rows. Display the Boolean result and count the duplicates.

```python
df[df.duplicated()]
```

#### Q2. Remove duplicate rows
Use drop_duplicates() to remove complete duplicate rows. Store the result in a new DataFrame and verify the remaining duplicate count.

```python
clean=df.drop_duplicates()
clean
```

#### Q3. Detect duplicate customer IDs
Identify records with repeated customer_id values. Display all records involved, including the first occurrence, using duplicated(keep=False).

```python
df[df['customer_id'].duplicated(keep=False)]
```

#### Q4. Remove duplicates using a subset
Remove duplicate records based on customer_id, keeping the first occurrence. Explain why this may be unsafe if two records with the same ID contain conflicting information.

```python
df = pd.DataFrame({
    "customer_id": [101, 102, 103, 102, 104, 105, 106, 106],
    "name": ["Asha", "rahul", "NEHA", "rahul",
             "Priya ", "Aman", "Riya", "Riya"],
    "age": ["25", " thirty", "35", " thirty",
            "29", "42", "28", "28"],
    "city": ["Delhi", "mumbai", "DELHI", "mumbai",
             " Pune ", "Delhi", None, None],
    "income": [30000, 50000, 45000, 50000,
               -1000, 70000, np.nan, np.nan]
})

df=df.drop_duplicates(subset='customer_id',keep='first')
df
```

#### Q5. Standardize text values
Clean the name and city columns by removing leading/trailing whitespace and standardizing letter case. Use string methods such as .str.strip(), .str.lower(), and .str.title().

```python
df['name']=df['name'].str.strip().str.lower()
df['city']=df['city'].str.strip().str.lower()
df[['name','city']]
```

#### Q6. Convert data types
Clean the age column by removing unnecessary whitespace and converting it to a numeric type. Invalid values such as "thirty" should become missing values rather than crashing the operation. Use pd.to_numeric() with errors="coerce".

```python
df['age']=pd.to_numeric(df['age'],errors="coerce")
df['age']
```

#### Q7. Identify invalid numerical values
Find records with negative income. Create a Boolean mask and display the invalid records.
Explain why a negative value should be investigated before automatically replacing it with zero.

```python
mask=df['income']>0
df.loc[~mask]
```

#### Q8. Replace invalid values
In a copy of the dataset, replace negative income values with NaN. Then report how many missing income values exist.
Use np.nan and verify the result with isna().

```python
df1=df.copy()
mask=df1['income']<0
df1.loc[mask,'income']=np.nan
df1['income']
```

```python
df1['income'].isna().sum()
```

#### Q9. Build a cleaning pipeline
Create a cleaned copy of df that:
1. Standardizes name and city.
2. Converts age to numeric, coercing invalid values to missing.
3. Replaces negative income values with missing values.
4. Removes exact duplicate rows.
5. Preserves the original DataFrame unchanged.
Check the final data types and missing-value counts.

```python
df = pd.DataFrame({
    "customer_id": [101, 102, 103, 102, 104, 105, 106, 106],
    "name": ["Asha", "rahul", "NEHA", "rahul",
             "Priya ", "Aman", "Riya", "Riya"],
    "age": ["25", " thirty", "35", " thirty",
            "29", "42", "28", "28"],
    "city": ["Delhi", "mumbai", "DELHI", "mumbai",
             " Pune ", "Delhi", None, None],
    "income": [30000, 50000, 45000, 50000,
               -1000, 70000, np.nan, np.nan]
})
df
```

```python
#Standardizes name and city.
df['name']=df['name'].str.strip().str.lower()
df['city']=df['city'].str.strip().str.lower()
df
```

```python
#Converts age to numeric, coercing invalid values to missing
df['age']=pd.to_numeric(df['age'],errors='coerce')
df
```

```python
#Replaces negative income values with missing values.
mask=df['income']<0
df.loc[mask,'income']=np.nan
df['income']
```

```python
df=df.drop_duplicates(keep='first')
df
```
